```python
def getTissueName(tissue_name):
    tissues = ['Brain-Anteriorcingulatecortex_BA24',
                 'Brain-Cortex',
                 'Brain-FrontalCortex_BA9',
                 'Brain-Putamen_basalganglia',
                 'Heart-AtrialAppendage',
                 'Liver',
                 'Lung',
                 'Muscle-Skeletal',
                 'Skin-NotSunExposed_Suprapubic',
                 'WholeBlood']
    tissues_old_name = ['Brain_Anterior_cingulate_cortex_BA24',
                         'Brain_Cortex',
                         'Brain_Frontal_Cortex_BA9',
                         'Brain_Putamen_basal_ganglia',
                         'Heart_Atrial_Appendage',
                         'Liver',
                         'Lung',
                         'Muscle_Skeletal',
                         'Skin_Not_Sun_Exposed_Suprapubic',
                         'Whole_Blood']
    
    idx = tissues_old_name.index(tissue_name)
    new_name = tissues[idx]
    return new_name
```

```python
bigwig_dir = '/project/yangili1/cfbuenabadn/leafcutter2_paper/code/resources/GTEx/BigWig'

with open('/project/yangili1/cfbuenabadn/leafcutter2_paper/code/results/ds_sashimi_plots/input/GTEx.bigwig_list.tsv', 'w') as fh:
    fh.write('sample\tbigwigPath\tGroup_label\tstrand\n')
    for tissue in tissues_old_name:
        tissue_files = sorted(os.listdir(f'/project/yangili1/cfbuenabadn/leafcutter2_paper/code/resources/GTEx/BigWig/{tissue}'))
        tissue_samples = [x.split('.')[0] for x in tissue_files if x.endswith('.bw')]
        tissue_new = getTissueName(tissue)
        for sample in tissue_samples:
            row_string = f'{sample}\t{bigwig_dir}/{tissue}/{sample}.bw\t{tissue_new}\t.\n'
            fh.write(row_string)
```

```python
def make_sashimi_links(cluster, psi, tissue_list):
    
    cluster_name = cluster.replace(':', '_')
    
    out_file = f'/project/yangili1/cfbuenabadn/leafcutter2_paper/code/results/ds_sashimi_plots/links/{cluster_name}.bed'
    
    with open(out_file, 'w') as fh:
    
        introns = psi.loc[psi.cluster == cluster]

        first_line = True
        header = '#Chrom\tstart\tend\tjunc\tgid\tstrand'
        
        for idx, row in introns.iterrows():
            intron_name = row.intron
            chrom, start, end, cluster = intron_name.split(':')
            strand = cluster.split('_')[-1]
            intron_row = [chrom, start, end, intron_name, cluster, strand]
            for tissue in tissue_list:
                tissue_intron_psi = str(round(float(row[tissue]*100), 2))

                tissue_files = sorted(os.listdir(f'/project/yangili1/cfbuenabadn/SpliFi/code/resources/GTEx/BigWig/{tissue}'))
                tissue_samples = [x.split('.')[0] for x in tissue_files if x.endswith('.bw')][:5]         
                psi_row = [tissue_intron_psi]*len(tissue_samples)
                intron_row.extend(psi_row)
                
                if first_line:
                    header += '\t' + '\t'.join(tissue_samples)

            if first_line:
                header += '\n'
                fh.write(header)
            row_to_write = '\t'.join(intron_row) + '\n'
            fh.write(row_to_write)
            first_line = False
```
