# Supplementary figures — the `Yang supplementary figures.ipynb` group

`SFig1` and `SFig3`. Both plot numbers that were recorded by hand in raw cells
of the source notebook — a LeafCutter2 timing benchmark and a junction-category
tally. There are no data files behind them, so unlike the other groups this
notebook has **no helpers module**: the constants below are the data.

| Panel | What it shows |
|---|---|
| `SFig1` | LeafCutter2 runtime vs number of junctions processed |
| `SFig3` | Unproductive splice junctions called by LeafCutter2 vs Gencode categories |

Panels go to `plots/`. Note both are saved as **`.svg` only** in the source
notebook, with no `dpi` — that is preserved here rather than silently promoted
to the png/pdf/svg triple the other groups use.

````python
import os

import numpy as np
from matplotlib import pyplot as plt

PLOTS_DIR = 'plots'
os.makedirs(PLOTS_DIR, exist_ok=True)
print('panels ->', PLOTS_DIR)
````

## `SFig1` — LeafCutter2 runtime

From a raw cell in the source notebook:

```
TIME: 20.495072 10000 junctions
TIME: 190.572404 100000 junctions
TIME: 491.135388 250000 junctions
TIME: 877.549837 500000 junctions
TIME: 1416.531623 750000 junctions
TIME: 1917.737630 1000000 junctions
```

````python
# SFig1 -- from Yang supplementary figures.ipynb
juncs_x = [1e4, 1e5, 2.5e5, 5e5, 7.5e5, 1e6]
time_y = np.array([20.495072, 190.572404, 491.135388, 877.549837, 1416.531623, 1917.737630])
min_y = time_y / 60

fig, ax = plt.subplots(figsize=(3, 3), dpi=300)
ax.scatter(juncs_x, min_y)
ax.plot(juncs_x, min_y)
ax.set_xlabel('Junctions processed')
ax.set_ylabel('Time (minutes)')
ax.set_xticks([1e4, 1e5, 2.5e5, 5e5, 7.5e5, 1e6],
              ['1e4', '1e5', '2.5e5', '5e5', '7.5e5', '1e6'], rotation=45)
ax.spines[['top', 'right']].set_visible(False)

plt.savefig(f'{PLOTS_DIR}/SFig1.svg', bbox_inches='tight')
print('wrote', f'{PLOTS_DIR}/SFig1.svg')

# Original: plt.savefig('../code/plots/SFig1.svg', bbox_inches='tight')
````

## `SFig3` — unproductive junctions, LeafCutter2 vs Gencode

From a raw cell in the source notebook:

```
269331 LeafCutter2-Unproductive
12472 Gencode-NonsenseMediatedDecay
8100 Gencode-CDSNotDefined
1733 Gencode-RetainedIntron
```

The fifth bar is the sum of the three Gencode categories (22,305).

````python
# SFig3 -- from Yang supplementary figures.ipynb
y = [269331, 12472, 8100, 1733, 12472 + 8100 + 1733]

fig, ax = plt.subplots(figsize=(3, 3), dpi=300)
ax.barh(range(5)[::-1], y)

y_str = ['269,331', '12,472', '8,100', '1,733', '22,305']
for i, y_ in enumerate(y[::-1]):
    ax.text(y_ + 12000, i - 0.1, str(y_str[-1 - i]))

ax.spines[['top', 'right']].set_visible(False)
ax.set_yticks(range(5), ['Gencode - all unproductive', 'Gencode - Retained intron',
                         'Gencode - CDS not defined',
                         'Gencode - Nonsense mediated decay', 'LeafCutter2 - Unproductive'])
ax.set_xlabel('Splice junctions per category')

plt.savefig(f'{PLOTS_DIR}/SFig3.svg', bbox_inches='tight')
print('wrote', f'{PLOTS_DIR}/SFig3.svg')

# Original: plt.savefig('../code/plots/SFig3.svg', bbox_inches='tight')
````
