```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
juncs_x = [1e4, 1e5, 2.5e5, 5e5, 7.5e5, 1e6]
time_y = np.array([20.495072, 190.572404, 491.135388, 877.549837, 1416.531623, 1917.737630])
min_y = time_y/60
```

```python
min_y
```

```python
fig, ax = plt.subplots(figsize=(3, 3), dpi=300)
ax.scatter(juncs_x, min_y)
ax.plot(juncs_x, min_y)
ax.set_xlabel('Junctions processed')
ax.set_ylabel('Time (minutes)')
ax.set_xticks([1e4, 1e5, 2.5e5, 5e5, 7.5e5, 1e6], ['1e4', '1e5', '2.5e5', '5e5', '7.5e5', '1e6'], rotation=45)
ax.spines[['top', 'right']].set_visible(False)

plt.savefig('../code/plots/SFig1.svg', bbox_inches='tight')
```
