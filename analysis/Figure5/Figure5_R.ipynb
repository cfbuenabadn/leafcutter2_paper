# Figure 5 | Impact of unproductive splicing on Alzheimer's disease -- R panels

R panels for **Figure 5**, matching the manuscript caption lettering.

| Panel | File | What it shows |
|---|---|---|
| **a** (left) | `Fig5A-left` | QQ plot of DLPFC **eQTL** signal among u-sQTLs vs p-sQTLs |
| **a** (right) | `Fig5A-right` | QQ plot of **pQTL** signal among u-sQTLs vs p-sQTLs |
| **b** | `Fig5B` | PTWAS + colocalization of u-sQTL effects with AD GWAS |

Panel **d** (right) is Python and lives in `Figure5_Python.ipynb`. Panels **c**
and **e**, and the LocusCompare half of **d**, are not built here.

Run **cell 3** once to load (and cache) the plot-ready data, then the plot
cells. On later sessions, uncomment **cell 2** and skip cell 3 to re-plot from
`figure_data/`. Panels are written to `plots/`.

`multiqq` draws its expected quantiles from `runif`, so the QQ panels are
stochastic. The seeds from `Fig5_R.ipynb` are kept: `set.seed(0)` for
`Fig5A-left`, `set.seed(3)` for `Fig5A-right`. Do not drop them.

Panel name changed to match the caption: `Fig5C_wide` -> `Fig5B`. The original
`ggsave` path is kept commented in each cell.

```r
source("Figure5_R_helpers.R")
source("Figure5_R_plot_helpers.R")

dir.create(PLOTS_DIR, showWarnings = FALSE, recursive = TRUE)
cat("panels will be written to", PLOTS_DIR, "\n")
```

```r
# Fast path: re-plot straight from the cache written by run_all().
# Uncomment this cell and skip the next one.

# data <- load_plot_data("figure_data")
# Fig5A_left <- data$Fig5A_left
# Fig5A_right <- data$Fig5A_right
# Fig5B   <- data$Fig5B
```

```r
# Reads the three RDS files from code/Ru_plots/ and caches them in figure_data/.

data <- run_all("figure_data")
Fig5A_left <- data$Fig5A_left
Fig5A_right <- data$Fig5A_right
Fig5B   <- data$Fig5B
```

```r
# Fig. 5a (left) | Fig5A-left -- from Fig5_R.ipynb
# QQ plot of DLPFC eQTL signal among u-sQTLs and p-sQTLs. u-sQTLs show stronger
# enrichment among eQTLs than their productive counterparts. Raw P values are
# from TensorQTL's nominal pass.
# Stochastic: set.seed(0) reproduces the published panel.

set.seed(0)
qq_plot <- plot_qq_panel(Fig5A_left, UP_PALETTE)
qq_plot

save_panel(qq_plot, "Fig5A-left", width = 3, height = 3)

# Original destination in Fig5_R.ipynb:
# ggsave("../code/plots/Fig5/Fig5A-left.svg", qq_plot, width = 3, height = 3, units = "in", dpi = 300)
```

```r
# Fig. 5a (right) | Fig5A-right -- from Fig5_R.ipynb
# The same comparison against protein expression QTLs (pQTL). u-sQTLs are again
# more enriched than p-sQTLs. Raw P values from TensorQTL's nominal pass.
# Stochastic: set.seed(3) reproduces the published panel.

set.seed(3)
qq_plot <- plot_qq_panel(Fig5A_right, UP_PALETTE)
qq_plot

save_panel(qq_plot, "Fig5A-right", width = 3, height = 3)

# Original destination in Fig5_R.ipynb:
# ggsave("../code/plots/Fig5/Fig5A-right.svg", qq_plot, width = 3, height = 3, units = "in", dpi = 300)
```

```r
# Fig. 5b | Fig5B -- from Fig5_R.ipynb  (was named Fig5C_wide)
# Integration of u-sQTL effects with AD GWAS through two-sided PTWAS (y axis)
# and colocalization (colour). Several signals were identified using u-sQTLs
# from multiple brain regions. Grey dashed line, P = 0.05; blue dashed line,
# the multiple-testing-corrected threshold based on the number of tests.
# Significant colocalization was defined as PP.H4 > 0.8.

Fig5B_plot <- plot_Fig5B(Fig5B, FIG5B_PALETTE)
print(Fig5B_plot)

save_panel(Fig5B_plot, "Fig5B", width = 14, height = 9)

# Original destination in Fig5_R.ipynb:
# ggsave("../code/plots/Fig5/Fig5B.svg", Fig5B_plot, width = 14, height = 9, units = "in", dpi = 300)
```
