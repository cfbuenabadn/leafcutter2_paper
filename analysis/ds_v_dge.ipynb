```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
psi = pd.read_csv('GTEx.psi.tsv.gz', sep='\t')
cluster_counts = pd.read_csv('GTEx.cluster_counts.tsv.gz', sep='\t')
psi_p = pd.read_csv('GTEx.psi_pvals.tsv.gz', sep='\t')
```

```python
ten_tissues =['Brain-Anteriorcingulatecortex_BA24',
 'Brain-Cortex',
 'Brain-FrontalCortex_BA9',
 'Brain-Putamen_basalganglia',
 'Heart-AtrialAppendage',
 'Liver',
 'Lung',
 'Muscle-Skeletal',
 'Skin-NotSunExposed_Suprapubic',
 'WholeBlood']
```

```python
psi['intron_cluster'] = psi.intron + ':' + psi.cluster + psi.gene_id#.loc[]#[ten_tissues]
```

```python
psi = psi.loc[psi.ctype=='PR,UP'].dropna()
```

```python
psi.index = psi.intron
```

```python
psi.loc[(psi[ten_tissues].quantile(0.9, axis=1) >= 0.25) & (psi[ten_tissues].quantile(0.3, axis=1) <= 0.05)]
```

```python
cluster_counts.index = cluster_counts.intron
cluster_counts = cluster_counts.loc[cluster_counts.ctype=='PR,UP'].dropna()
```

```python
cluster_counts = cluster_counts.groupby(['cluster'])[ten_tissues].sum()
```

```python
min_per_tissue= pd.DataFrame(cluster_counts.min(axis=1))
min_per_tissue.columns = ['min_cluster_counts']
```

```python
psi = pd.merge(psi, min_per_tissue, right_on='cluster', left_on='cluster')
```
